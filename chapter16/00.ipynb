```python
# 파일 전체를 한 줄씩 처리하려면 파일 객체를 반복합니다.

with open("memo.txt", "r", encoding="utf-8") as file: # 이 영역을 벗어나면 파일객체 사용불가. 그래서 안전한것
    for idx, line in enumerate(file, start=1):
        print(f"{idx}. {line.strip()}") # 한 라인을 읽을 때 strip은 필수
```

```python
with open("memo.txt", "r", encoding="utf-8") as file: # 이 영역을 벗어나면 파일객체 사용불가. 그래서 안전한것
    seq = 1
    for line in file:
        print(f"{seq}번째 라인: {line.strip()}")
        seq+=1
```

```python
with open("result.txt", "w", encoding="utf-8") as file:

    file.write("파이썬 파일 쓰기 실습\n")

    file.write("두 번째 줄입니다.\n")
```

```python
with open("result.txt", "a", encoding="utf-8") as file:

    file.write("새로운 내용을 추가합니다.\n")
```

```python
# 새로운 내용으 추가합니다.1
# ...
# 새로운 내용을 추가합니다.10
# 마지막 라인에는 \n을 추가하지 말아주세요
with open("result.txt", "a", encoding="utf-8") as file:
    for seq in range(10):
        if seq != 9:
            
            file.write(f"새로운 내용을 추가합니다.{seq+1} \n")
        else:
            file.write(f"새로운 내용을 추가합니다.{seq+1}")
```
