```python
H, W = map(int, input().split()) # 공백으로 구분하여 이미지의 높이와 너비를 받음

# 2차원 밝기 리스트 생성
brightness_matrix = [] 

for _ in range(H):
    row = [sum(map(int, pixel.split(','))) // 3 for pixel in input().split()]
    brightness_matrix.append(row)

# 2차원 리스트의 모든 요소를 다 더해서 임계값 계산
total_sum = sum(sum(row) for row in brightness_matrix)
threshold = total_sum // (H * W)

# 4. 임계값 출력 및 흑백 이미지 행 단위 출력
print(threshold)
for row in brightness_matrix:
    # 각 픽셀 밝기가 threshold 이상이면 '1', 미만이면 '0'으로 변환 후 이어서 출력 ex. 101, 000, 111
    print("".join("1" if brightness >= threshold else "0" for brightness in row))
```
