```python
def greet(name, subject):
    print(f"안녕하세요! {name}님")
    print(f"{subject} 공부를 시작합니다.")

greet("홍길동", "파이썬") # 파이썬
greet("채희주", "sql") # sql
```

```python
# 두 개의 숫자를 입력 받아서 더한 값을 반환(return)하는 함수

# 함수 이름: first_sum
def first_sum(num1, num2):
    return num1 + num2

int_num1 = int(input("첫번째 숫자: "))
int_num2 = int(input("두번째 숫자: "))

print(f"결과 : {first_sum(int_num1, int_num2)}")
```

```python
# 두 수를 입력받아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오.
# 4개 결과값을 한번에 반환
# 사용자가 입력한 값이 숫자가 아닐 경우 오류 메시지 출력하고 프로그램 종료

def calc(num1, num2):
    length= len(num1) * len(num2)
    if length !=1:
        return "한 자리 수만 입력하세요"
    elif not(57 >= ord(num1) >= 49 and 57 >= ord(num2) >= 49):
        return "값이 정수가 아닙니다."
    num1 = int(num1)
    num2 = int(num2)
    return num1 + num2, num1 - num2, num1 * num2, num1 / num2

int_num1 = input("첫번째 숫자(1~9): ")
int_num2 = input("두번째 숫자(1~9): ")

print(calc(int_num1, int_num2))

```
