```python
class Student: #실행을 해야 쓸 수 있음

    def __init__(self, name, score, age): # 생성자. class가 생성될 때 생성된다.

        self.name = name
        self.score = score
        self.age = age
    
    def show_result(self):
        print((f"{self.name}: {self.score}"))

    def is_passed(self):
        return self.score >= 60
```

```python
student1 = Student("홍길동", 75, 5)
student2 = Student("강감찬", 55, 5)
student3 = Student("권율", 61, 5)

students = [student1, student2, student3] # 객체 리스트

for student in students:
    print(f"{student.name}님의 점수는 {student.score}점 입니다.")

    if student.is_passed():
        result = '당당히 합격'
    else:
        result = '아쉽지만 탈락'

    print(f"{student.name}님은 이번 시험에서 {result} 하셨습니다.")
```
